```python
# PART - A 
# DICTIONARIES
## 1. CREATE AN EMPTY DICTIONARY
inventory = {}
print(inventory)
```

```python
## 2. STORE THE FIRST PRODUCT DETAILS IN THE VARIABLES
# ### Product 1
productName1 = "Mobile Phone"
productQuantity1 = 5
productPrice1 = 20000
productReleaseYear1 = 2020
```

```python
## 3. ADD DETAILS IN INVENTORY
inventory["ProductNo1"] = {
    "Product Name": productName1,
    "Product Quantity": productQuantity1,
    "Product Price": productPrice1,
    "Product Release Year": productReleaseYear1
}
    
```

```python
## 4. STORE THE SECOND PRODUCT DETAILS IN VARIABLES

# ### Product 2
productName2 = "Laptop"
productQuantity2 = 3
productPrice2 = 50000
productReleaseYear2 = 2022
```

```python
## 5. ADD THE SECOND PRODUCT DETAILS IN THE INVENTORY
inventory["ProductNo2"] = {
    "Product Name": productName2,
    "Product Quantity": productQuantity2,
    "Product Price": productPrice2,
    "Product Release Year": productReleaseYear2
}
```

```python
## 6. DISPLAY THE PRODUCTS PRESENT IN THE INVENTORY
print(inventory)
```

```python
## 7. CHECK IF RELEASE YEARS ARE PRESENT IN INVENTORY

if "Product Release Year" in inventory["ProductNo1"]:
    print("ProductNo1 release year is present")
else:
    print("ProductNo1 release year not present")

if "Product Release Year" in inventory["ProductNo2"]:
    print("ProductNo2 release year is present")
else:
    print("ProductNo2 release year not present")
```

```python
## 8. DELETE RELEASE YEAR OF BOTH PRODUCTS
del inventory["ProductNo1"]["Product Release Year"]
del inventory["ProductNo2"]["Product Release Year"]

print("Inventory after deleting release years:")
print(inventory)
```

```python
## 1. CREATE A TUPLE OF PRODUCT PRICES
prices = (250,300,150,400,100,350,200)
print(prices)
```

```python
## 2. FIND AND PRINT THE HIGHEST AND LOWEST PRICE
high_price = max(prices)
low_price = min(prices)
print("highest price is ",high_price)
print("lowest price is",low_price)
```

```python
## 3. calculate and print the total sum
total_sum = sum(prices)
print(total_sum)
```

```python
## 4. CONVERT TUPLE INTO SORTED LIST
sorted_prices = sorted(prices)
print(sorted_prices)
```
