```python
#Problem 1: Create a DataFrame
import pandas as pd

data = {
    'Student': ['David', 'Samuel', 'Terry', 'Evan'],
    'Age': [27, 24, 22, 32],
    'Country': ['UK', 'Canada', 'China', 'USA'],
    'Course': ['Python', 'Data Structures',
               'Machine Learning', 'Web Development'],
    'Marks': [85, 72, 89, 76]
}

df = pd.DataFrame(data)

print(df)
```

```python
#Problem 2: Retrieve the Marks column and assign it to variable b
b = df['Marks']

print(b)
```

```python
#Problem 3: Retrieve Country and Course columns and assign them to variable c
c = df[['Country', 'Course']]

print(c)
```

```python
#Problem 4: Data Loading and Exploration
import pandas as pd

data = {
    'Customer_ID': [101, 102, 103, 104, 105,
                    106, 107, 108, 109, 110],
    'Name': ['Aman', 'Priya', 'Rahul', 'Neha', 'Arjun',
             'Simran', 'Karan', 'Pooja', 'Rohan', 'Anjali'],
    'Age': [25, 32, 28, 45, 35, 29, 41, 31, 26, 38],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male',
               'Female', 'Male', 'Female', 'Male', 'Female'],
    'Purchase_Amount': [2500, 3200, 1800, 4500, 3000,
                        2200, 5000, 3500, None, 4000]
}

df = pd.DataFrame(data)

## Display the first 5 rows
print(df.head())

## Get summary statistics
print(df.describe())

## Check missing values
print(df.isnull().sum())

## Handle missing Purchase_Amount using the column mean
df['Purchase_Amount'] = df['Purchase_Amount'].fillna(
    df['Purchase_Amount'].mean()
)

## Check missing values again
print(df.isnull().sum())
```

```python
#Problem 5: Data Filtering and Selection
##A. Retrieve customers whose Age is greater than 30
customers_above_30 = df[df['Age'] > 30]

print(customers_above_30)
```

```python

##B. Calculate total Purchase Amount for male and female customers separately
total_purchase_by_gender = df.groupby('Gender')['Purchase_Amount'].sum()

print(total_purchase_by_gender)
```

```python
##C. Select only Customer_ID, Name, and Purchase_Amount
selected_columns = df[
    ['Customer_ID', 'Name', 'Purchase_Amount']
]

print(selected_columns)
```

```python
#Problem 6: Data Transformation and Aggregation
##A. Add a new column Discounted_Price with a 10% discount
df['Discounted_Price'] = df['Purchase_Amount'] * 0.90

print(df)
```

```python
##B. Group the data by Gender and calculate the average Purchase Amount
average_purchase_by_gender = df.groupby('Gender')[
    'Purchase_Amount'
].mean()

print(average_purchase_by_gender)
```

```python
##C. Sort the dataset by Purchase_Amount in descending order
sorted_df = df.sort_values(
    by='Purchase_Amount',
    ascending=False
)

print(sorted_df)
```
