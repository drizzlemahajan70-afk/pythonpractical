```python
# SHOPPING LIST PROGRAM

## 1. Create an Empty Shopping List

shopping_list = []
```

```python
## 2. Add Items to the Shopping List

### Adding Watch, Laptop, Shoes, Pen and Clothes

shopping_list = ["Watch", "Laptop", "Shoes", "Pen", "Clothes"]

print("Shopping List:", shopping_list)
```

```python
## 3. Add a New Item

### Adding Football to the Shopping List

shopping_list.append("Football")

print("After adding Football:", shopping_list)
```

```python
## 4. Print the Last Item

### Displaying the Last Item of the List

print("Last item:", shopping_list[-1])
```

```python

## 5. Print the Entire Shopping List

### Displaying All Items

print("Entire Shopping List:", shopping_list)
```

```python

## 6. Print Important Items

### Displaying Laptop and Shoes

print("Important items:", shopping_list[1:3])
```

```python
## 7. Change an Item

### Replacing Pen with Notebook

shopping_list[3] = "Notebook"

print("After changing Pen to Notebook:", shopping_list)
```

```python
## 8. Delete the Least Important Item

### Removing Watch from the Shopping List

shopping_list.remove("Watch")

print("Final Shopping List:", shopping_list)
```
