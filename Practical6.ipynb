```python
# PROBLEM 1 : BUGGY CODE
# FIX THE ABOVE PROBLEM WITH SUPER IN PYTHON
class Person:
    def __init__(self, name, id):
        self.name = name
        self.id = id
    def Display(self):
        print(self.name, self.id)


class Emp(Person):

    def __init__(self, name, id):
        
        super().__init__(name, id)

    def Print(self):
        print("Emp class called")

Emp_details = Emp("MEERA", 103)
Emp_details.Display()
Emp_details.Print()
```

```python
# PROBLEM 2 - WRITE A PYTHON PROGRAM TO FIND ANAGRAM CHECK.
str1 = input("Enter first string: ")
str2 = input("Enter second string: ")
str1 = str1.lower()
str2 = str2.lower()

if sorted(str1) == sorted(str2):
    print("Strings are Anagrams")
else:
    print("Strings are not Anagrams")
```

```python
# PROBLEM 3 - WRITE A PYTHON PROGRAM TO CHECK FOR ARMSTRONG NUMBER
num = int(input("Enter a number: "))
original = num
digits = len(str(num))
sum = 0
while num > 0:
    digit = num % 10
    sum = sum + digit ** digits
    num = num // 10

if sum == original:
    print("It is an Armstrong number")
else:
    print("It is not an Armstrong number")
```

```python
# PROBLEM 4 - DEAL WITH CLASSES AND OBJECTS
class Vehicle:                                           #TASK 1
    color = "white"                                      #TASK 2
    def __init__(self, max_speed, mileage):
        self.max_speed = max_speed
        self.mileage = mileage
        self.seating_capacity = 0
        
    def assign_seating_capacity(self, seating_capacity): #TASK 3
        self.seating_capacity = seating_capacity

    def display_properties(self):                        #TASK 4
        print("Maximum Speed:", self.max_speed, "kmph")
        print("Mileage:", self.mileage, "kmpl")
        print("Color:", self.color)
        print("Seating Capacity:", self.seating_capacity)

car1 = Vehicle(200, 20)
car1.assign_seating_capacity(5)

car2 = Vehicle(180, 25)
car2.assign_seating_capacity(4)

print("Car 1 Details:")                                    #TASK 5
car1.display_properties()                        

print()

print("Car 2 Details:")
car2.display_properties()
```
