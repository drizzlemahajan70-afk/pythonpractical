```python
# EXERCISE 1 - HANDLING ZERO DIVISION ERROR
def safe_divide(numerator, denominator):
    try:
        result = numerator / denominator
        return result
    except ZeroDivisionError:
        print("Error: Cannot divide by Zero.")
        return None

num1 = float(input("Enter numerator: "))
num2 = float(input("Enter denominator: "))

print("Result:", safe_divide(num1, num2))

```

```python
# EXERCISE 2 - HANDLING VALUE ERROR
import math

def square_root(number1):
    try:
        if number1 < 0:
            raise ValueError
        return math.sqrt(number1)
    except ValueError:
        print("Invalid input! Please enter a positive integer or a float value.")
        return None

num = float(input("Enter a number: "))

print("Square root:", square_root(num))
```

```python
# EXERCISE - 3 HANDLING GENERIC EXCEPTIONS
def calculate(num):
    try:
        result = num / (num - 5)
        return result
    except Exception as e:
        print("An error occurred during calculation.")
        return None

num = float(input("Enter a number: "))

print("Result:", calculate(num))
```
