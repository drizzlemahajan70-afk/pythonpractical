```python
#Problem 1: Zipper File()
##Question: Given a filename and a list of strings, create a file where every two consecutive strings are combined (zippered)
##          and written on separate lines.

def zipper_file(filename, strings):
    with open(filename, 'w') as file:
        for i in range(0, len(strings), 2):
            file.write(strings[i] + strings[i+1] + '\n')

    print("File created successfully!")



strings = ["Hello", "World", "Python", "Programming"]


zipper_file("output.txt", strings)


with open("output.txt", 'r') as file:
    print(file.read())
```

```python
#Problem 2: Reformat Date()
## Question: Given a list of dates in MM/DD/YYYY format, convert them into the format:

##Office Hours on Month DD, YYYY


def reformat_date(dates):
    months = {
        "01": "January",
        "02": "February",
        "03": "March",
        "04": "April",
        "05": "May",
        "06": "June",
        "07": "July",
        "08": "August",
        "09": "September",
        "10": "October",
        "11": "November",
        "12": "December"
    }

    result = []

    for date in dates:
        month, day, year = date.split('/')
        formatted_date = f"Office Hours on {months[month]} {day}, {year}"
        result.append(formatted_date)

    return result


## Example
dates = ["01/15/2024", "03/09/2025", "12/25/2026"]

print(reformat_date(dates))
```
