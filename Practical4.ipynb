```python
# 4. WRITE A PYTHON PROGRAM IN A JUPYTER NOTEBOOK THAT TAKES AAN INTEGER INPUT (1 TO 9999) FROM THE USER AND
# DETERMINES THE FOLLOWING USING BRANCING STATEMENTS:
num = int(input("Enter an integer between 1 and 9999: "))

if num < 1 or num > 9999:
    print("enter a number between 1 and 9999.")
else:
    # Check even or odd
    if num % 2 == 0:
        print("The number is even.")
    else:
        print("The number is odd.")

    # Check palindrome
    original = num
    reverse = 0

    while num > 0:
        digit = num % 10
        reverse = reverse * 10 + digit
        num = num // 10

    if original == reverse:
        print("The number is a palindrome.")
    else:
        print("The number is not a palindrome.")

    # Calculate sum of digits
    temp = original
    digit_sum = 0

    while temp > 0:
        digit_sum = digit_sum + temp % 10
        temp = temp // 10

    # Check divisibility
    if original % digit_sum == 0:
        print("The number is divisible by the sum of its digits.")
    else:
        print("The number is not divisible by the sum of its digits.")
```

```python
# 7. WRITE A WHILE LOOP TO COPY THE STRINGS 'ORANGE' OF THE LIST TO THE LIST NEW_SQAURES.
# STOP AND EXIT THE LOOP IF THE VALUE ON THE LIST IS NOT 'ORANGE'
squares = ['red', 'yellow', 'green', 'purple', 'blue']

new_squares = []

i = 0

while i < len(squares):
    if squares[i] != 'orange':
        break

    new_squares.append(squares[i])
    i += 1

print("New squares:", new_squares)
```

```python
# 8. THE FOLLOWING IS A LIST OF ANIMALS IN A NATIONAL ZOO.
# ANIMALS = ["LION","GIRAFFE","GORILLA","PARROTS","CROCODILE","DEER","SWAN"]
# YOUR BROTHER NEEDS TO WRITE AN ESSAY ON THE AMINALS WHOSE NAME ARE MADE OF 7 LETTERS .
# HELP HIM FIND THOSE ANIMALS THROUGH A WHILE LOOP AND CREATE A SEPARATE LIST OF SUCH ANIMALS.
Animals = ["lion", "giraffe", "gorilla", "parrots", "crocodile", "deer", "swan"]

seven_letter_animals = []

i = 0

while i < len(Animals):
    if len(Animals[i]) == 7:
        seven_letter_animals.append(Animals[i])

    i += 1

print("Animals with 7 letters:")
print(seven_letter_animals)
```
